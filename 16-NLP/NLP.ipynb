## O que é Processamento de Linguagem Natural (NLP)?

É uma pequena área de inteligência artificial que lida com as interações entre humanos e computadores usando linguagem natural. NLP é conhecida por desenvolver algorítmos e modelos computacionais capazes de levar máquinas a entenderem, analisarem e replicar o idioma humano.

image.png

É um campo multidisciplinar que se baseia em técnicas da ciência da computação, letras, matemática e psicologia. Sua missão é construir sistemas que podem processar e entender o idioma humano, que é complexo e diverso, envolvendo várias etapas até o entendimento.

### Aplicações do NLP

- Classificação de Texto: envolve a categorização de dados textuais, ou seja, é a aplicação de categorias ou classes já existentes aos dados textuais em mãos. É útil para detectar mensagens spam e analisar sentimentos.

- Reconhecimento de entidade nomeada (NER): identifica e extrai entidades, como pessoas, organizações e localizações dos dados de texto, sendo útil para a obtenção de informações de artigos e publicações nas redes sociais.

- Parte do discurso (PSO): categoriza cada palavra de uma frase, como substantivo, verbo, adjetivo ou advérbio. É uma ferramenta crucial, pois capacita máquinas a entenderem a estrutura gramatica de frases e identificar as funções de cada palavra na frase. Uma categorização gramatical bem feita pode melhorar a performance da classificação de textos, tradução de máquina e análise de sentimentos.

- Modelagem de linguagem: constrói modelos probabilísticos de linguagem, com o objetivo de prever a semelhança entre sequências de palavras diferentes.

- Tradução de máquina: desenvolve sistemas que aumtomaticamente traduzem o texto de uma língua para outra, sendo útil para negócios internacionais e comunicação entre culturas diferentes.

- Recuperação de informação: responsável por obter informações pertinentes de uma coleção vasta de dados textuais.

- Sistemas de resposta a perguntas: constrói sistemas automatizados que podem responder a perguntas em linguagem natural. É aplicável em serviços aos clientes e em ambientes educacionais.

- Resumo de texto: transforma um longo texto em um resumo, como um artigo ou documento.

- Geração de texto: sistemas criativos que geram um textos com linguagem natural, como chatbots ou assistente de escrita.

- Reconhecimento de fala: converte uma linguagem de fala para texto.

### Principais bibliotecas

- Natural Language Toolkit (NLTK): disponibiliza uma gama de ferramentas para tarefas diversas, como tokenização, categorização de classe gramatica, análise, análise de sentimentos, etc.

- spaCy: nova biblioteca, código aberto, tem alta performance e é eficiente ao analisar grandes dados de texto. Efetua a tokenização, marcação de classe gramatica, análise, reconhecimento de entidade, etc.

##### Separação de Frases

É a divisão do texto em partes únicas

```python
def segmentando_texto(texto):
    frases = []
    inicio = 0

    for i in range(len(texto)):
        if texto[i] == '.' or texto[i] == '!' or texto[i] == '?':
            frases.append(texto[inicio:i+1].strip())
            inicio = i + 1

    return frases
```

##### Tokenização: Tokeniza o texto em palavras, frases ou sentenças.

```python
def tokenizar(s):
  palavras = [] #token words should be stored here
  i = 0
  palavra = ""
  while(i <len(s)):
    if (s[i] != " "):
      palavra = palavra+s[i]
    else:
        palavras.append(palavra)
        palavra = ""

    i = i + 1
  palavras.append(palavra)
  return palavras
```

##### Stop words

São palavras com pouco significado, com "o", "a", "os", "as", "que", etc. São geralmente artigos, preposições ou conjunções. Elas não são úteis, precisam ser removidas para um desenvolvimento melhor do modelo, evitando ruídos.

```python
def remover_stop_words(palavras):
    j = 0
    while (j < len(palavras)):
        if(len(palavras[j]) <= 3):
            palavras.remove(palavras[j])
        else:
            j = j + 1

    return palavras
```

Podemos também personalizar o comportamento de detecção de frases utilizando delimitadores customizados. Aqui utilizamos reticências (...) como nosso delimitador, e também o ponto final (.):

```python
ellipsis_text = (
    "Gus, can you, ... never mind, I forgot"
    " what I was saying. So, do you think"
    " we should ..."
)
```

```python
from spacy.language import Language
```

```python
@Language.component("set_custom_boundaries")
def set_custom_boundaries(doc):
  """Add support to use `...` as a delimiter for sentence detection"""
  for token in doc[:-1]:
    if token.text == "...":
      doc[token.i + 1].is_sent_start = True
      return doc
```

```python
custom_nlp = spacy.load("en_core_web_sm")
custom_nlp.add_pipe("set_custom_boundaries", before="parser")
custom_ellipsis_doc = custom_nlp(ellipsis_text)
custom_ellipsis_sentences = list(custom_ellipsis_doc.sents)
```

```python
for sentence in custom_ellipsis_sentences:
  print(sentence)
```

Neste exemplo, você usou o decorador @Language.component("set_custom_boundaries") para definir uma nova função que obtém um objeto Doc como argumento. O trabalho dessa função é achar os tokens do Doc que estão nos inícios das frases e marcar seu atributo .is_sent_start para True. Uma vez feito isso, a função deve retornar o objeto Doc novamente.

Então, você pode adicionar funções limitadoras personalizadas ao objeto Language utilizando o método .add_pipe(). A análise de texto com esse objeto Language alterado agora irá tratar a palavra soente após as reticências como o início da frase seguinte.

Como vários aspectos do spaCy, podemos personalizar o processo de tokenização para detectar token de palavras específicas. Isso é geralmente usado para palavras com hífen.

Para customizar a tokenização, precisaremos atualizar a propriedade tokenizer no objeto Language com um novo objeto tokenizer.

```python
custom_about_text = (
    "Gus Proto is a Python developer currently"
     " working for a London@based Fintech"
     " company. He is interested in learning"
     " Natural Language Processing."
)
print([token.text for token in nlp(custom_about_text)[8:15]])
```

Vemos aqui que o padrão de leitura informou que London@based é um único token, mas se usássemos um hífen em vez do símbolo @, teríamos 3 tokens (London, - e based).

Para dizer que o símbolo @ é um infixo personalizado, precisaremos construir nosso próprio objeto tokenizer:

```python
import re
from spacy.tokenizer import Tokenizer
```

```python
import spacy
from spacy.tokenizer import Tokenizer

# Carregar o modelo padrão do spaCy
custom_nlp = spacy.load("en_core_web_sm")

# Compilar as regexes padrão de prefixos e sufixos
prefix_re = spacy.util.compile_prefix_regex(custom_nlp.Defaults.prefixes)
suffix_re = spacy.util.compile_suffix_regex(custom_nlp.Defaults.suffixes)

# Adicionar o infixo personalizado
custom_infixes = list(custom_nlp.Defaults.infixes) + [r"@"]
infix_re = spacy.util.compile_infix_regex(custom_infixes)

# Definir o tokenizer personalizado
custom_nlp.tokenizer = Tokenizer(
    custom_nlp.vocab,
    prefix_search=prefix_re.search,
    suffix_search=suffix_re.search,
    infix_finditer=infix_re.finditer,
    token_match=None,
)

# Processar o texto com o tokenizer personalizado
custom_tokenizer_about_doc = custom_nlp(custom_about_text)

# Imprimir os tokens do intervalo especificado
print([token.text for token in custom_tokenizer_about_doc[8:15]])
```

Neste exemplo, primeiro instanciamos um novo objeto linguístico. Para construir um novo Tokenizer, você geralmente cria ele com:
- Vocab: Um contâiner que armazena casos especiais, usados para lidar como contrações e emoticons
- prefix_search: é uma função que processa pontuações prévias, como a abertura de parênteses
- suffix_search: lida com pontuação sucetiva, como o fechamento de parênteses
- infix_finditer: trabalha com separadores não vazios, como hífens
- token_match: é uma função opcional que junta strings que não deveriam ser separadas. Ela substitui as regras antigas e é útil para entidades como URLs ou números.

As funções envolvidas aqui são geralmente funções que você pode acessar dos objetos regex compilados. Para construir objetos regex para prefixos e sufixos você pode gerá-los com os padrões.

Para a construção de uma função infixo personalizada, primeiro você define uma lista com qualquer parentesco regex que você queira incluir. Então, você junta sua lista customizada com o atributo do objeto Language.Defaults.infixes, que precisa ser convertido em uma lista antes de uní-la. Então você leva a tupla extendida como um argumento até o spacy.util.compile_infix_regex() para obter seu novo objeto regex para infixos.

Quando você chama o construtor Tokenizar, você passa o método .search() no prefixo e o sufixo como objetos regex, e a função .finditer() no objeto regex infixo. Agora você pode substituir o tokenizador com o objeto custom_nlp, vendo que o símbolo @ foi tokenizado separadamente.

### Lematização

A lematização é o processo de redução das formas flexionadas de uma palavra, garantindo ao mesmo tempo que a forma reduzida pertence ao idioma. Esta forma reduzida, ou palavra raíz, é o lema.

Por exemplo: organiza, organizado e organizando são formas de organiza. Aqui, organiza é o lema. A inflexão da palavra permite que você expresse categorias gramaticais diferentes, como tempo verbal (organiza x organizado), quantidade (pessoa x pessoas) e assim por diante. A lematização é necessária pois ajuda a reduz as formas flexionadas de uma palavra, para que possam ser analsiadas como um único ítem. Isso pode ajudar a normalizar o texto.

SpaCy coloca o atributo lemma_ na classe do Token. Esse atrbuto leva o Token a sua forma lematizada:

```python
text_3 = "Gus is helping organize a developer conference on Applications of Natural Language Processing. He keeps organizing local Python meetups and several internal talks at his workplace."
```

```python
conference_help_doc = nlp(text_3)
```

```python
for token in conference_help_doc:
  if str(token) != str(token.lemma_):
    print(f"{str(token):>20} : {str(token.lemma_)}")
```

Neste exemplo, você faz a verificação para ver se a palavra riginal é diferente do lema, e se for, mostramos ambas.

Você notará, por exemplo, que o o rganize reduz a forma do lema. Se você não lematizar o texto, então o organize e o organizing serão contados como tokens diferentes, mesmo que ambos se refiram ao mesmo conceito. Lematização ajuda a evitar palavras duplicadas que podem sobrepor conceitualmente.

### Correspondência baseada em regras

É uma das formas de extrair informações de um texto não estruturado. É usado para identificar e extrair tokens de frases de acordo com os padrões, como letras minúsculas, e características gramaticais, como uma parte da fala.

Você pode usar expressões gerais para extrair as entidades, como números de telefones, mas a correspondência baseada em regras é mais forte que o regex sozinho, pois você pode incluir filtros semânticos ou gramaticais.

Por exemplo, com a correspondência baseada em regras você pode extrair o nome e sobrenome de alguém, pois são sempre pronomes próprios:

```python
about_doc = nlp(text_2)
```

```python
from spacy.matcher import Matcher
```

```python
matcher = Matcher(nlp.vocab)
```

```python
def extract_full_name(nlp_doc):
  pattern = [{"POS": "PROPN"}, {"POS": "PROPN"}]
  matcher.add("FULL_NAME", [pattern])
  matches = matcher(nlp_doc)
  for _, start, end in matches:
    span = nlp_doc[start:end]
    yield span.text
next(extract_full_name(about_doc))
```

Nesse exemplo, o pattern (padrão) é uma lista de objetos que define a combinação de tokens que corresponderam. As tags POS dentro dele são PROPn (pronomes próprios). Então, o pattern consiste em dois objetos o qual as tags POS dos dois devem ser PROPN. Esse padrão é então adicionado ao Matcher com o método .add(), que leva um identificador chave e uma lista de padrões. Finalmente, os correspondentes são obtidos com seus índices de início e fim.

Podemos usar correspondência baseada em regras para extrair números de telefones:

```python
text_7 = ("There is a developer conference"
     " happening on 21 July 2019 in London. It is titled"
     ' "Applications of Natural Language Processing".'
     " There is a helpline number available"
     " at (123) 456-7891")
```

```python
def extract_phone_number(nlp_doc):
    # Crie o objeto Matcher
    matcher = Matcher(nlp_doc.vocab)

    # Defina o padrão para o número de telefone
    pattern = [
        {"ORTH": "("},
        {"SHAPE": "ddd"},
        {"ORTH": ")"},
        {"SHAPE": "ddd"},
        {"ORTH": "-", "OP": "?"},
        {"SHAPE": "dddd"}
    ]

    # Adicione o padrão ao Matcher
    matcher.add("PHONE_NUMBER", [pattern])

    # Aplique o Matcher ao documento
    matches = matcher(nlp_doc)

    # Itere sobre os resultados e retorne o número de telefone
    for match_id, start, end in matches:
        span = nlp_doc[start:end]
        return span.text
```

```python
conference_org_doc = nlp(text_7)
```

```python
extract_phone_number(conference_org_doc)
```

Neste exemplo, o padrão foi atualizado para corresponder aos números do telefone. Aqui, alguns atributos do token foram usados:
- ORTH: corresponde ao texto exato do token
- SHAPE: transforma a string do token para mostrar características ortográficas, d são dígitos
- OP: define os operadores. Usando ? como um valor significa que o padrão é opcional, dizendo que pode corresponder 0 ou 1 vez.

Encadear esses dicionários oferece muita flexibilidade para escolher os critérios de correspondência.

Novamente, a correspondência baseada em regras ajuda a identificar e extrair tokens e frases, combinando de acordo com padrões lexicais e características gramaticais. Isso pode ser útil quando você procura uma entidade específica.

```python
def is_token_allowed(token):
  return bool(
         token
         and str(token).strip()
         and not token.is_stop
         and not token.is_punct
     )
```

Um caso de uso do NER é redigir nomes de pessoas de um texto. Por exemplo, você pode querer fazer isso para ocultar informações pessoais coletadas em uma pesquisa. Dê uma olhada no seguinte exemplo: Pitão

```python
survey_text = (
     "Out of 5 people surveyed, James Robert,"
     " Julie Fuller and Benjamin Brooks like"
     " apples. Kelly Cox and Matthew Evans"
     " like oranges."
 )
```

```python
def replace_person_names(token):
  if token.ent_iob != 0 and token.ent_type_ == "PERSON":
    return "[REDACTED] "
  return token.text_with_ws
```

```python
def redact_names(nlp_doc):
  with nlp_doc.retokenize() as retokenizer:
    for ent in nlp_doc.ents:
      retokenizer.merge(ent)
  tokens = map(replace_person_names, nlp_doc)
  return "".join(tokens)
```

```python
survey_doc = nlp(survey_text)
```

```python
print(redact_names(survey_doc))
```

Neste exemplo, replace_person_names() usa .ent_iob, que fornece o código IOB da tag de entidade nomeada usando marcação inside-outside-beginning (IOB).

A função redact_names() usa um retokenizer para ajustar o modelo de tokenização. Ele obtém todos os tokens e passa o texto por map() para substituir quaisquer tokens de destino por [REDIGIDO].

Assim, você seria capaz de redigir uma grande quantidade de texto em segundos, enquanto fazê-lo manualmente poderia levar muitas horas. Dito isso, é preciso sempre ter cuidado com a redação, pois os modelos não são perfeitos!

### Filtrando Stop Words

Palavras irrelevantes são palavras que você deseja ignorar, então você as filtra do seu texto ao processá-lo. Palavras muito comuns como 'in', 'is' e 'an' são frequentemente usadas como palavras irrelevantes, uma vez que não acrescentam muito significado a um texto por si só.

Veja como importar as partes relevantes do NLTK para filtrar palavras irrelevantes:

```python
nltk.download("stopwords")
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
worf_quote = "Sir, I protest. I am not a merry man!"
```

Agora tokenize worf_quote por palavra e armazene a lista resultante em words_in_quote:

```python
words_in_quote = word_tokenize(worf_quote)
words_in_quote
```

Você tem uma lista de palavras em worf_quote, então a próxima etapa é criar um conjunto de palavras irrelevantes para filtrar palavras_em_quote. Para este exemplo, você precisará se concentrar nas palavras irrelevantes em "inglês":

```python
stop_words = set(stopwords.words("english"))
```

Em seguida, crie uma lista vazia para conter as palavras que passam pelo filtro:

```python
filtered_list = []
```

Você criou uma lista vazia, filtered_list, para conter todas as palavras em words_in_quote que não são palavras irrelevantes. Agora você pode usar stop_words para filtrar words_in_quote:


```python
for word in words_in_quote:
   if word.casefold() not in stop_words:
        filtered_list.append(word)
```

Você iterou palavras_in_quote com um loop for e adicionou todas as palavras que não eram palavras irrelevantes à lista_filtrada. Você usou .casefold() no Word para poder ignorar se as letras do Word eram maiúsculas ou minúsculas. Vale a pena fazer isso porque stopwords.words('english') inclui apenas versões em minúsculas de palavras irrelevantes.

Alternativamente, você pode usar uma compreensão de lista para fazer uma lista de todas as palavras em seu texto que não são palavras irrelevantes:


```python
filtered_list = [word for word in words_in_quote if word.casefold() not in stop_words]
```

Ao usar uma compreensão de lista, você não cria uma lista vazia e depois adiciona itens ao final dela. Em vez disso, você define a lista e seu conteúdo ao mesmo tempo. Usar uma compreensão de lista costuma ser visto como mais pitônico.

Você filtrou algumas palavras como 'sou' e 'a', mas também filtrou 'não', o que afeta o significado geral da frase. (Worf não ficará feliz com isso.)

Palavras como “eu” e “não” podem parecer importantes demais para serem filtradas e, dependendo do tipo de análise que você deseja fazer, podem ser. Aqui está o porquê:
- 'Eu' é um pronome, que são palavras de contexto em vez de palavras de conteúdo:
  - As palavras de conteúdo fornecem informações sobre os tópicos abordados no texto ou o sentimento que o autor tem sobre esses tópicos.
  - Palavras de contexto fornecem informações sobre o estilo de escrita. Você pode observar padrões em como os autores usam palavras de contexto para quantificar seu estilo de escrita. Depois de quantificar seu estilo de escrita, você pode analisar um texto escrito por um autor desconhecido para ver até que ponto ele segue um estilo de escrita específico, para tentar identificar quem é o autor.
- 'not' é tecnicamente um advérbio, mas ainda foi incluído na lista de palavras irrelevantes do NLTK para o inglês. Se quiser editar a lista de palavras irrelevantes para excluir 'não' ou fazer outras alterações, você pode baixá-la.

### Tags nas Partes de Fala

Parte do discurso é um termo gramatical que trata dos papéis que as palavras desempenham quando você as usa juntas em frases. Marcar classes gramaticais, ou marcação POS, é a tarefa de rotular as palavras em seu texto de acordo com sua classe gramatical.

Em inglês, existem oito classes gramaticais:

Substantivo: É uma pessoa, lugar ou coisa
- montanha, bagel, Polônia

Pronome: Substitui um substantivo
- você, ela, nós

Adjetivo: Fornece informações sobre como é um substantivo
- eficiente, ventoso, colorido

Verbo: É uma ação ou um estado de ser
- aprender, é, vá

Advérbio: Fornece informações sobre um verbo, um adjetivo ou outro advérbio
- de forma eficiente, sempre, muito

Preposição: Fornece informações sobre como um substantivo ou pronome está conectado a outra palavra
- from, about, at

Conjunção: Conecta duas outras palavras ou frases
- assim, porque, e

Interjeição: É uma exclamação
- sim, ai, uau

Algumas fontes também incluem a categoria artigos (como “a” ou “the”) na lista de classes gramaticais, mas outras fontes os consideram adjetivos. NLTK usa a palavra determinante para se referir a artigos.

É assim que você pode identificar entidades nomeadas! Mas você pode dar um passo adiante e extrair entidades nomeadas diretamente do seu texto. Crie uma string da qual extrair entidades nomeadas. Você pode usar esta citação de A Guerra dos Mundos:


```python
quote = "Men like Schiaparelli watched the red planet—it is odd, by-the-bye, that for countless centuries Mars has been the star of war—but failed to interpret the fluctuating appearances of the markings they mapped so well. All that time the Martians must have been getting ready. During the opposition of 1894 a great light was seen on the illuminated part of the disk, first at the Lick Observatory, then by Perrotin of Nice, and then by other observers. English readers heard of it first in the issue of Nature dated August 2."
```

Agora crie uma função para extrair entidades nomeadas:


```python
def extract_ne(quote):
     words = word_tokenize(quote, language='english')
     tags = nltk.pos_tag(words)
     tree = nltk.ne_chunk(tags, binary=True)
     return set(
         " ".join(i[0] for i in t)
         for t in tree
         if hasattr(t, "label") and t.label() == "NE"
     )
```

Com esta função você reúne todas as entidades nomeadas, sem repetições. Para fazer isso, você tokeniza por palavra, aplica tags de classe gramatical a essas palavras e, em seguida, extrai entidades nomeadas com base nessas tags. Como você incluiu binary=True, as entidades nomeadas que você obterá não serão rotuladas de forma mais específica. Você apenas saberá que eles são entidades nomeadas.

Dê uma olhada nas informações que você extraiu:

```python
extract_ne(quote)
```

Você perdeu a cidade de Nice, possivelmente porque o NLTK a interpretou como um adjetivo normal em inglês, mas ainda obteve o seguinte:
-Uma instituição: 'Observatório Lick'

-Um planeta: 'Marte'
